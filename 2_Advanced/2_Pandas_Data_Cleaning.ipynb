```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt  

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df.loc[:10, 'salary_year_avg': 'salary_hour_avg']
```

### Extrat the median for `salary_year_avg` & `salary_hour_avg`

```python
df['salary_year_avg'].median()
```

```python
df['salary_hour_avg'].median()
```

#### Made a variable for thoes median i.e `median_salary_year` & `median_salary_hour`

```python
median_salary_year = df['salary_year_avg'].median()
```

```python
median_salary_hour = df['salary_hour_avg'].median()
```

### Filled thoes missing or null value with the generated median value

```python
df_filled = df

df_filled['salary_year_avg'] =  df_filled['salary_year_avg'].fillna(median_salary_year)  
df_filled['salary_hour_avg'] =  df_filled['salary_hour_avg'].fillna(median_salary_hour)
```

```python
df_filled.loc[:10,  'salary_year_avg' : 'salary_hour_avg']
```

## Here we are learning how to removw duplicates value
***DataFrame.drop_duplicates(subset = None, *, keep = 'first', inplace = False, ignore_index = False)***

```python
df_unique = df_filled

df_unique = df_unique.drop_duplicates()

print('Len of org df:              ', len(df_filled))
print('Len of Drop Duplicates df:  ', len(df_unique))
print('Rows Dropped:               ', len(df_filled)-len(df_unique))
```

```python
df_unique = df_unique.drop_duplicates(subset=['job_title', 'company_name'])

print('Len of org df:              ', len(df_filled))
print('Len of Drop Duplicates df:  ', len(df_unique))
print('Rows Dropped:               ', len(df_filled)-len(df_unique))
```
