#Ploting Job Posting Dates


```python
# Install datasets Library (if not already installed)
# !pip install -U datasets

# Importing Libraries
import pandas as pd
from datasets import load_dataset

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df.job_posted_date.head()
```

```python
plt.plot(df.job_posted_date, df.job_posted_date)
plt.show()
```

```python
date_count = df.job_posted_date.value_counts()
date_count = date_count.sort_index()
```

```python
plt.plot(date_count.index, date_count.values)
plt.show()
```

```python
df['job_posted_monnth'] = df['job_posted_date'].dt.month
```

```python
df
```

```python
month_count = df.job_posted_monnth.value_counts()
month_count = month_count.sort_index()
```

```python
month_count
```

```python
plt.plot(month_count.index, month_count.values)
plt.show()
```
