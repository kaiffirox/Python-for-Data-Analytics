```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from datasets import load_dataset

# Simple plot
x = np.arange(0, 5, 0.1)
y = np.sin(x)
plt.plot(x, y)
plt.show()

# Load limited data
dataset = load_dataset('lukebarousse/data_jobs', split='train[:100000]')
df = dataset.to_pandas()

# Date cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])

# Monthly aggregation
df['job_posted_month'] = df['job_posted_date'].dt.to_period('M')
month_count = df['job_posted_month'].value_counts().sort_index()

plt.plot(month_count.index.astype(str), month_count.values)
plt.xticks(rotation=45)
plt.show()

# Top job titles
job_count = df['job_title'].value_counts().head(10)
job_count
```

```python
job_count = df.job_title.value_counts().head(10)
```

```python
plt.bar(job_count.index,  job_count)
```

```python
job_count = job_count.sort_values(ascending=True)
plt.barh(job_count.index,  job_count)
plt.show()
```

#***Matplotlib : Labeling***

```python
job_count = job_count.sort_values(ascending=False)
plt.bar(job_count.index,  job_count)
plt.title('Top 10 Job Titles')
plt.xlabel('Job Title')
plt.ylabel('Count of Job Posting')
plt.xticks(rotation=45, ha='right')
plt.show()
#
plt
plt.show()
```
